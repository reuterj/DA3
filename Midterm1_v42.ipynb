<b>(4 points) Question 3a.</b> <i>Please use a simulation with 10,000 iterations to estimate the average future value and expected utility associated with the savings rate and asset allocation described below. Store these values in a list or tuple named `q3a` and print `q3a`.</i>

- You will invest 15% of your annual income in a retirement account that invests 100% in small-cap equity, with an expected return of 12%, standard deviation of 30%, and minimum possible net return of -100%. As in class, I describe this asset as having returns N(0.12,0.30).
- Your annual income will remain constant at \\$100,000 and you be investing 15% of your income today and at the beginning of years 1 through 29 (for a total of 30 contributions). Each year you will earn a random return on your existing account balance plus your new contribution.
- Assume your utility is given by the square root of your retirement account balance 30 years from today.
- I always recommend starting a simulation by coding a single trial, in this case the utility from investing 10% of your income in the asset for 30 years based on one set of thirty small-cap equity returns.
- <b>Note: If run into any problem, please simplify the problem until you are able to get the simulation to run and then add comments explaining the changes you need to make to answer this question.</b>

```python
# your code goes here
```

```python
# 1 trial

bal = 0
for j in range(30):
    bal = (bal+15000)*(1+max(random.gauss(0.12,0.30),-1))
util = math.sqrt(bal)
bal, util
```

```python
# 10000 trials

trials = []
for i in range(10000):
    bal = 0
    for j in range(30):
        bal = (bal+15000)*(1+max(random.gauss(0.12,0.30),-1))
    util = math.sqrt(bal)
    trials.append([bal,util])
    
q3a = statistics.mean([t[0] for t in trials]), statistics.mean([t[1] for t in trials])
print(q3a)
```
