```python
def most_common(s):
    """Identifies the most common character(s) in a string
       and returns those character(s) in a list"""
    parsed    = [[s.count(char), char] for char in set(s)]
    max_count = max([element[0] for element in parsed])
    sub       = [element[1] for element in parsed if element[0] == max_count]
    return sub
```

### Finance Formulas

Cumulative n-period net return $ = (1+r_1)(1+r_2) \cdots (1+r_n) - 1$

Geometric annual n-period net return $ = \left( (1+r_1)(1+r_2) \cdots (1+r_n) \right)^{\frac{1}{n}} - 1$

$PV = \frac{FV_n}{(1 + r)^n}$ where $FV_n$ is cash flow $n$ periods in the future and $r$ is appropriate per-period rate

To calculate $NPV$, you sum the PV of each cash flow associated with a project, including the initial outlay.

$FV_n = PV (1+r)^n$ where $n$ is number of periods in the future and $r$ is the appropriate per-period rate. Note that $(1+r)^n$ is a cumulative n-period gross return.

$PV(annuity) = AnnuityFactor \times PMT = \left[ \frac{1}{r} - \frac{1}{r(1+r)^n} \right] \times PMT$ where $r$ is the per period rate, $n$ is the number of time periods, $PMT$ is the per-period payment, and the first payment is made one period from today


```python
# Effective annual rate with continuous compounding when APR = 7%

math.exp(0.07)-1
```

```python
def cr(l):
    """Calculates cumulative net return taking list of net returns as input"""
    return math.prod([1+element for element in l]) - 1
```

```python
def geo(l):
    """Calculates geometric average return taking list of net returns as input"""
    return math.prod([1+element for element in l])**(1/len(l)) - 1
```

```python
def af(r,n):
    """annuity factor where r is per-period rate and n is number periods"""
    return (1/r) - (1/(r*((1+r)**n)))
```

```python
# EU Simulation

# Option A: Invest $10,000 in N(0.12,0.18)
# Option B: Pay broker $250 and invest $9750 in N(0.12,0.14). 

# Investment horizon 10 years
# Minimum possible net return is -99.99%
# Utility is natural log of ending portfolio value.

# Which option has the higher certainty equivalent?
```

```python
port1 = []
port2 = []

for num in range(10000):
    r1 = [ max(random.normalvariate(1.12,0.18),0.0001) for num in range(10) ]
    r2 = [ max(random.normalvariate(1.12,0.14),0.0001) for num in range(10) ]

    fv1 = math.prod(r1)*10000
    fv2 = math.prod(r2)* 9750

    port1.append(fv1)
    port2.append(fv2)
```

```python
plt.hist(port1, bins=100)
plt.hist(port2, bins=100, color='r', alpha=0.5)
plt.show()
```

```python
ln_port1 = [math.log(element) for element in port1]
ln_port2 = [math.log(element) for element in port2]

eu1 = statistics.mean(ln_port1)
eu2 = statistics.mean(ln_port2)

ce1 = math.exp(eu1)
ce2 = math.exp(eu2)
```

```python
print(f'Mean balance:        {statistics.mean(port1):8.2f}  {statistics.mean(port2):8.2f}')
print(f'Stdev balance:       {statistics.stdev(port1):8.2f}  {statistics.stdev(port2):8.2f}')
print(f'Expected Utility:    {eu1:8.2f}  {eu2:8.2f}')
print(f'Certainty Equiv.:    {ce1:8.2f}  {ce2:8.2f}')
```

```python
# I was asked to add code comparing certainty equivalent of buy-and-hold versus rebalancing
# but once you have two sets of portfolio values, you can use cells above...

port1 = []
port2 = []

for i in range(10000):
    ret_e = [max(random.gauss(1.12, 0.18), 0.0001) for y in range(10)]
    ret_b = [max(random.gauss(1.05, 0.07), 0.0001) for y in range(10)]
    fv_a = (0.6*math.prod(ret_e) + 0.4*math.prod(ret_b))*10000
    ret = [0.6*re + 0.4*rb for re, rb in zip(ret_e, ret_b)]
    fv_b = math.prod(ret)*10000
    port1.append(fv_a)
    port2.append(fv_b)
```

```python
# Goal: Create function to calculate certainty equivalent (CE)
#       Calculates CE of future values when investing 'amount' for one year
#       Assumes asset has annual return 'mu' and annual std dev 'sd'
#       Assumes investor utility measured by natural log of future value

def ce(mu,sd,amount):
    fv = [amount * (1 + max(random.gauss(mu,sd),-0.9999)) for num in range(10000)]
    ut = [math.log(element) for element in fv]
    eu = statistics.mean(ut)
    return math.exp(eu)
```
